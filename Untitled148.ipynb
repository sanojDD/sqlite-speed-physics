```python
import sqlite3
import time
import os

DB_FILE = "physics_benchmark.db"

def cleanup():
    """Removes old database files to ensure a fresh test environment."""
    for f in [DB_FILE, DB_FILE + "-wal", DB_FILE + "-shm"]:
        if os.path.exists(f):
            os.remove(f)

print("🚀 Starting Database Performance Physics Benchmark...\n")

# ==========================================
# 1. BASELINE: Default Rollback Journal
# ==========================================
cleanup()
conn = sqlite3.connect(DB_FILE)
conn.execute("CREATE TABLE accounts (id INTEGER PRIMARY KEY, balance REAL)")
conn.commit()

count = 1000
start = time.time()
for i in range(count):
    conn.execute("INSERT INTO accounts (id, balance) VALUES (?, ?)", (i, 100.0))
    conn.commit() # Individual fsync per transaction
duration = time.time() - start
print(f"1. Baseline (Rollback Journal):    {count / duration:,.2f} TPS")
conn.close()


# ==========================================
# 2. WAL MODE: Write-Ahead Logging
# ==========================================
cleanup()
conn = sqlite3.connect(DB_FILE)
conn.execute("PRAGMA journal_mode=WAL;")
conn.execute("CREATE TABLE accounts (id INTEGER PRIMARY KEY, balance REAL)")
conn.commit()

start = time.time()
for i in range(count):
    conn.execute("INSERT INTO accounts (id, balance) VALUES (?, ?)", (i, 100.0))
    conn.commit()
duration = time.time() - start
print(f"2. WAL Mode Enabled:              {count / duration:,.2f} TPS")
conn.close()


# ==========================================
# 3. SYNCHRONOUS SETTINGS (FULL, NORMAL, OFF)
# ==========================================
for sync_setting in ["FULL", "NORMAL", "OFF"]:
    cleanup()
    conn = sqlite3.connect(DB_FILE)
    conn.execute("PRAGMA journal_mode=WAL;")
    conn.execute(f"PRAGMA synchronous={sync_setting};")
    conn.execute("CREATE TABLE accounts (id INTEGER PRIMARY KEY, balance REAL)")
    conn.commit()

    start = time.time()
    for i in range(count):
        conn.execute("INSERT INTO accounts (id, balance) VALUES (?, ?)", (i, 100.0))
        conn.commit()
    duration = time.time() - start
    print(f"3. Synchronous = {sync_setting:<6}       {count / duration:,.2f} TPS")
    conn.close()


# ==========================================
# 4 & 5. BATCHING & LARGER BATCH SIZES
# ==========================================
total_rows = 50000
for batch_size in [100, 1000, 10000, 50000]:
    cleanup()
    conn = sqlite3.connect(DB_FILE)
    conn.execute("PRAGMA journal_mode=WAL;")
    conn.execute("PRAGMA synchronous=NORMAL;")
    conn.execute("CREATE TABLE accounts (id INTEGER PRIMARY KEY, balance REAL)")
    conn.commit()

    start = time.time()
    # Group commits: executing thousands of inserts inside a single transaction block
    for chunk_start in range(0, total_rows, batch_size):
        chunk_end = min(chunk_start + batch_size, total_rows)
        with conn:
            for i in range(chunk_start, chunk_end):
                conn.execute("INSERT INTO accounts (id, balance) VALUES (?, ?)", (i, 100.0))
    duration = time.time() - start
    print(f"4/5. Batch Size = {batch_size:<6}        {total_rows / duration:,.2f} TPS")
    conn.close()


# ==========================================
# 6. CPU BOTTLENECK (In-Memory / No Disk I/O)
# ==========================================
# By using an in-memory database, we completely eliminate disk lag
# to see the strict limit imposed by Python and CPU execution cycles.
conn = sqlite3.connect(":memory:")
conn.execute("PRAGMA synchronous=OFF;")
conn.execute("CREATE TABLE accounts (id INTEGER PRIMARY KEY, balance REAL)")
conn.commit()

mem_rows = 100000
start = time.time()
with conn:
    for i in range(mem_rows):
        conn.execute("INSERT INTO accounts (id, balance) VALUES (?, ?)", (i, 100.0))
duration = time.time() - start
print(f"6. CPU Bottleneck (In-Memory):    {mem_rows / duration:,.2f} TPS")
conn.close()

print("\n✨ Benchmark complete! Notice how TPS scales from hundreds to hundreds of thousands as disk and sync overhead are removed.")
```

```python

import sqlite3
import time
import random
import os
import math
import statistics

# =========================
# 1. CONFIGURATION
# =========================
ACCOUNTS = 10_000
INITIAL_BALANCE_CENTS = 100_000  # $1,000.00 per account
DURATION_SEC = 5
REPEATS = 2
DB_FILE = "transfer_benchmark.db"

# Compare different SQLite storage/durability settings
CONFIGS = [
    ("DELETE", "FULL"),
    ("WAL", "FULL"),
    ("WAL", "NORMAL"),
    ("WAL", "OFF"),
]

DEBIT_SQL = """
    UPDATE accounts
    SET balance = balance - ?
    WHERE id = ? AND balance >= ?
"""

CREDIT_SQL = """
    UPDATE accounts
    SET balance = balance + ?
    WHERE id = ?
"""


# =========================
# 2. CLEANUP
# =========================
def cleanup():
    for path in (DB_FILE, DB_FILE + "-wal", DB_FILE + "-shm"):
        if os.path.exists(path):
            os.remove(path)


# =========================
# 3. CREATE AND SEED DATABASE
# =========================
def create_database(journal_mode, sync_mode):
    cleanup()

    conn = sqlite3.connect(DB_FILE)

    actual_mode = conn.execute(
        f"PRAGMA journal_mode={journal_mode}"
    ).fetchone()[0].upper()

    conn.execute(f"PRAGMA synchronous={sync_mode}")

    conn.execute("""
        CREATE TABLE accounts (
            id INTEGER PRIMARY KEY,
            balance INTEGER NOT NULL CHECK (balance >= 0)
        )
    """)

    with conn:
        conn.executemany(
            "INSERT INTO accounts (id, balance) VALUES (?, ?)",
            (
                (account_id, INITIAL_BALANCE_CENTS)
                for account_id in range(1, ACCOUNTS + 1)
            ),
        )

    return conn, actual_mode


# =========================
# 4. ONE ATOMIC MONEY TRANSFER
# =========================
def transfer(conn, from_id, to_id, amount_cents):
    try:
        # Start a write transaction before reading/updating balances.
        conn.execute("BEGIN IMMEDIATE")

        debit = conn.execute(
            DEBIT_SQL,
            (amount_cents, from_id, amount_cents),
        )

        # No row updated = insufficient funds or invalid source account.
        if debit.rowcount != 1:
            conn.rollback()
            return False

        credit = conn.execute(
            CREDIT_SQL,
            (amount_cents, to_id),
        )

        if credit.rowcount != 1:
            raise RuntimeError("Destination account does not exist")

        conn.commit()
        return True

    except Exception:
        if conn.in_transaction:
            conn.rollback()
        raise


# =========================
# 5. LATENCY HELPERS
# =========================
def percentile(values, p):
    if not values:
        return 0.0

    ordered = sorted(values)
    index = max(0, math.ceil(p * len(ordered)) - 1)
    return ordered[index]


# =========================
# 6. RUN ONE BENCHMARK
# =========================
def run_benchmark(journal_mode, sync_mode, repeat_number):
    conn, actual_mode = create_database(journal_mode, sync_mode)

    rng = random.Random(12345)

    attempted = 0
    successful = 0
    failed = 0
    latencies_ms = []

    print(
        f"Running {actual_mode}/{sync_mode}, "
        f"repeat {repeat_number}..."
    )

    start = time.perf_counter()
    deadline = start + DURATION_SEC

    while time.perf_counter() < deadline:
        from_id = rng.randint(1, ACCOUNTS)
        to_id = rng.randint(1, ACCOUNTS)

        while to_id == from_id:
            to_id = rng.randint(1, ACCOUNTS)

        # Amount between $1.00 and $50.00
        amount_cents = rng.randint(100, 5000)

        tx_start = time.perf_counter()
        success = transfer(conn, from_id, to_id, amount_cents)
        tx_end = time.perf_counter()

        attempted += 1
        latencies_ms.append((tx_end - tx_start) * 1000)

        if success:
            successful += 1
        else:
            failed += 1

    elapsed = time.perf_counter() - start

    # =========================
    # 7. CHECK DATA CORRECTNESS
    # =========================
    account_count, total_balance, minimum_balance = conn.execute("""
        SELECT COUNT(*), SUM(balance), MIN(balance)
        FROM accounts
    """).fetchone()

    expected_balance = ACCOUNTS * INITIAL_BALANCE_CENTS

    integrity_ok = (
        account_count == ACCOUNTS
        and total_balance == expected_balance
        and minimum_balance >= 0
    )

    # =========================
    # 8. WAL CHECKPOINT OBSERVATION
    # =========================
    wal_before = 0
    checkpoint_result = None
    wal_after = 0

    if actual_mode == "WAL":
        wal_path = DB_FILE + "-wal"

        if os.path.exists(wal_path):
            wal_before = os.path.getsize(wal_path)

        # Copy committed WAL content back to the database file
        checkpoint_result = conn.execute(
            "PRAGMA wal_checkpoint(TRUNCATE)"
        ).fetchone()

        if os.path.exists(wal_path):
            wal_after = os.path.getsize(wal_path)

    result = {
        "journal": actual_mode,
        "sync": sync_mode,
        "repeat": repeat_number,
        "seconds": elapsed,
        "attempted": attempted,
        "successful": successful,
        "failed": failed,
        "attempt_tps": attempted / elapsed,
        "successful_tps": successful / elapsed,
        "avg_ms": statistics.mean(latencies_ms) if latencies_ms else 0,
        "p50_ms": percentile(latencies_ms, 0.50),
        "p95_ms": percentile(latencies_ms, 0.95),
        "p99_ms": percentile(latencies_ms, 0.99),
        "integrity_ok": integrity_ok,
        "wal_before_kb": wal_before / 1024,
        "wal_after_kb": wal_after / 1024,
        "checkpoint": checkpoint_result,
    }

    conn.close()
    return result


# =========================
# 9. RUN ALL CONFIGURATIONS
# =========================
all_results = []

print("Starting SQLite transfer benchmark\n")

for journal_mode, sync_mode in CONFIGS:
    for repeat_number in range(1, REPEATS + 1):
        result = run_benchmark(
            journal_mode,
            sync_mode,
            repeat_number,
        )

        all_results.append(result)

        print(
            f"  Attempt TPS: {result['attempt_tps']:,.0f} | "
            f"Successful TPS: {result['successful_tps']:,.0f} | "
            f"Failed: {result['failed']:,}"
        )
        print(
            f"  Avg latency: {result['avg_ms']:.3f} ms | "
            f"p50: {result['p50_ms']:.3f} ms | "
            f"p95: {result['p95_ms']:.3f} ms | "
            f"p99: {result['p99_ms']:.3f} ms"
        )
        print(
            f"  Money/account check: "
            f"{'PASS' if result['integrity_ok'] else 'FAIL'}"
        )

        if result["journal"] == "WAL":
            print(
                f"  WAL size before checkpoint: "
                f"{result['wal_before_kb']:.1f} KB"
            )
            print(
                f"  WAL size after checkpoint:  "
                f"{result['wal_after_kb']:.1f} KB"
            )
            print(f"  Checkpoint result: {result['checkpoint']}")

        print()

# =========================
# 10. SUMMARY USING MEDIAN TPS
# =========================
print("=" * 75)
print("FINAL SUMMARY (MEDIAN ACROSS REPEATS)")
print("=" * 75)
print(
    f"{'Mode':<15} {'Attempt TPS':>14} "
    f"{'Success TPS':>14} {'p99 ms':>12} {'Integrity':>12}"
)

for journal_mode, sync_mode in CONFIGS:
    matching = [
        r for r in all_results
        if r["journal"] == journal_mode and r["sync"] == sync_mode
    ]

    attempt_tps = statistics.median(
        r["attempt_tps"] for r in matching
    )
    success_tps = statistics.median(
        r["successful_tps"] for r in matching
    )
    p99 = statistics.median(r["p99_ms"] for r in matching)
    integrity = all(r["integrity_ok"] for r in matching)

    print(
        f"{journal_mode + '/' + sync_mode:<15} "
        f"{attempt_tps:>14,.0f} "
        f"{success_tps:>14,.0f} "
        f"{p99:>12.3f} "
        f"{'PASS' if integrity else 'FAIL':>12}"
    )

print("=" * 75)
print("Benchmark finished.")
print("Note: OFF is for disposable benchmark data only.")
```

```python
import sqlite3
import time
import os
import random
import statistics

DB_FILE = "batch_transfer_benchmark.db"
ACCOUNTS = 10_000
INITIAL_BALANCE = 100_000
TOTAL_OPERATIONS = 50_000
BATCH_SIZES = [1, 100, 1000, 5000, 10000]

def cleanup():
    for f in [DB_FILE, DB_FILE + "-wal", DB_FILE + "-shm"]:
        if os.path.exists(f):
            os.remove(f)

results = []

print("SQLite Batch Transfer Benchmark")
print("=" * 88)
print(f"{'Batch':>8} {'Attempts':>10} {'Success':>10} "
      f"{'Duration(s)':>12} {'Transfers/sec':>16} {'Integrity':>12}")
print("-" * 88)

for batch_size in BATCH_SIZES:
    cleanup()
    conn = sqlite3.connect(DB_FILE)

    conn.execute("PRAGMA journal_mode=WAL")
    conn.execute("PRAGMA synchronous=NORMAL")

    conn.execute("""
        CREATE TABLE accounts (
            id INTEGER PRIMARY KEY,
            balance INTEGER NOT NULL
        )
    """)

    conn.executemany(
        "INSERT INTO accounts VALUES (?, ?)",
        ((i, INITIAL_BALANCE) for i in range(1, ACCOUNTS + 1))
    )
    conn.commit()

    rng = random.Random(42)
    successful = 0
    start = time.perf_counter()

    try:
        for chunk_start in range(0, TOTAL_OPERATIONS, batch_size):
            current_batch = min(
                batch_size, TOTAL_OPERATIONS - chunk_start
            )

            with conn:
                for _ in range(current_batch):
                    from_id = rng.randint(1, ACCOUNTS)
                    to_id = rng.randint(1, ACCOUNTS - 1)
                    if to_id >= from_id:
                        to_id += 1

                    amount = rng.randint(100, 1000)

                    cursor = conn.execute(
                        """UPDATE accounts
                           SET balance = balance - ?
                           WHERE id = ? AND balance >= ?""",
                        (amount, from_id, amount)
                    )

                    if cursor.rowcount == 1:
                        conn.execute(
                            "UPDATE accounts SET balance = balance + ? WHERE id = ?",
                            (amount, to_id)
                        )
                        successful += 1

        duration = time.perf_counter() - start

        count, total, minimum = conn.execute("""
            SELECT COUNT(*), SUM(balance), MIN(balance)
            FROM accounts
        """).fetchone()

        integrity = (
            count == ACCOUNTS
            and total == ACCOUNTS * INITIAL_BALANCE
            and minimum >= 0
        )

        tps = successful / duration

        results.append({
            "batch": batch_size,
            "success": successful,
            "duration": duration,
            "tps": tps,
            "integrity": integrity
        })

        print(f"{batch_size:>8,} {TOTAL_OPERATIONS:>10,} "
              f"{successful:>10,} {duration:>12.3f} "
              f"{tps:>16,.0f} "
              f"{'PASS' if integrity else 'FAIL':>12}")

    finally:
        conn.close()

cleanup()

print("-" * 88)
print("TPS = successful transfers per second")
print("Integrity checks: account count, total balance, and non-negative balances.")
print("Settings: WAL + synchronous=NORMAL")
```
